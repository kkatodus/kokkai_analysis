```python
import os
import re
from params.paths import ROOT_DIR
from file_handling.file_read_writer import read_json, write_json, create_dir

## URLS
GIJI_URL = "https://kokkai.ndl.go.jp/api/speech?"

UPPER_REPR_LIST_URL = "https://www.sangiin.go.jp/japanese/joho1/kousei/giin/218/giin.htm"
LOWER_REPR_LIST_URL = "https://www.shugiin.go.jp/internet/itdb_annai.nsf/html/statics/syu/1giin.htm"

UPPER_MEETING_INFO_PAGE_URL = "https://www.sangiin.go.jp/japanese/kon_kokkaijyoho/index.html"
LOWER_MEETING_INFO_PAGE_URL = "https://www.shugiin.go.jp/internet/itdb_iinkai.nsf/html/iinkai/list.htm"


## PATHS
UPPER_OUTPUT_DIR = os.path.join(ROOT_DIR,'data', "data_sangiin")
LOWER_OUTPUT_DIR = os.path.join(ROOT_DIR,'data', "data_shugiin")
if not os.path.exists(UPPER_OUTPUT_DIR):
    raise Exception("Upper house output directory does not exist")
if not os.path.exists(LOWER_OUTPUT_DIR):
    raise Exception("Lower house output directory does not exist")
if not os.path.exists(os.path.join(UPPER_OUTPUT_DIR, "repr_list")):
    print(os.listdir(UPPER_OUTPUT_DIR))
    raise Exception("Upper house repr list directory does not exist")
if not os.path.exists(os.path.join(LOWER_OUTPUT_DIR, "repr_list")):
    raise Exception("Lower house repr list directory does not exist")
```

```python
sangiin_repr_list_file_name = "20250911_repr_list.json"
shugiin_repr_list_file_name = "20250911_repr_list.json"
```

```python
import pandas as pd
import os

def convert(input_path:str, output_path:str, excel_path:str=None)->None:
	repr_list = read_json(input_path)

	# instantiate pandas empty dataframe
	df = pd.DataFrame(columns=["選挙区","候補者名(漢字）", "候補者名(ひらがな）", "所属政党",])

	for party_name in repr_list["reprs"]:
		for repr in repr_list["reprs"][party_name]:
			df = pd.concat([df, pd.DataFrame([{
				"選挙区": repr["district"],
				"候補者名(漢字）": repr["name"],
				"候補者名(ひらがな）": repr["yomikata"],
				"所属政党": party_name,
			}])], ignore_index=True)
	df.to_csv(output_path, index=False)
	if excel_path:
		df.to_excel(excel_path, index=False)

	
```

```python
convert(os.path.join(UPPER_OUTPUT_DIR, "repr_list", sangiin_repr_list_file_name), os.path.join(UPPER_OUTPUT_DIR, "repr_list", sangiin_repr_list_file_name.split(".")[0] + ".csv"), excel_path=os.path.join(UPPER_OUTPUT_DIR, "repr_list", sangiin_repr_list_file_name.split(".")[0] + ".xlsx"))
convert(os.path.join(LOWER_OUTPUT_DIR, "repr_list", shugiin_repr_list_file_name), os.path.join(LOWER_OUTPUT_DIR, "repr_list", shugiin_repr_list_file_name.split(".")[0] + ".csv"), excel_path=os.path.join(LOWER_OUTPUT_DIR, "repr_list", shugiin_repr_list_file_name.split(".")[0] + ".xlsx"))
```
